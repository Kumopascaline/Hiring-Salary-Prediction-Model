### Stage 1: Setup and Libraries

In this stage, we install `word2number` to handle text-based data (like "five") and import the necessary tools for data handling and machine learning.

```
# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/hiring (1).csv')

# Preview the data to see missing values (NaN)
print("Original Data:")
print(df)
```

### Stage 3: Data Preprocessing (Cleaning)

Machine learning models cannot handle missing values or text strings. So we fill the gaps and convert words to numbers.

```
print("Information about the datasets")
print(df.info())
print("-------------------------------------------------------------------------------------------------------")
print("Checking for null values")
print(df.isnull().sum())
print("-------------------------------------------------------------------------------------------------------")
print("Checking for duplicate values")
print(df.duplicated().sum())
```

```
# 1. Fill missing experience values with 'zero'
df.experience = df.experience.fillna("zero")

# 2. Convert words like 'five' to 5
df.experience = df.experience.apply(w2n.word_to_num)

# 3. Fill missing test scores with the median value of that column
median_test_score = df['test_score(out of 10)'].median()
df['test_score(out of 10)'] = df['test_score(out of 10)'].fillna(median_test_score)

print("Cleaned Data:")
print(df)
```

### Stage 4: Training the Linear Regression Model

Defining our features (Experience, Test Score, Interview Score) and our target (Salary), then training the model to find the mathematical relationship between them.

```
# Initialize the Linear Regression model
reg = linear_model.LinearRegression()

# Train (fit) the model using the cleaned data
# X = independent variables, y = target variable (salary)
reg.fit(df[['experience', 'test_score(out of 10)', 'interview_score(out of 10)']], df['salary($)'])

print("Model training complete.")
```

### Stage 5: Predicting Future Salaries

Now we use the trained model to predict salaries for the new candidates given

```
print("----------------------------------------------------------- Predictions --------------------------------------------------------------------------------------")

# Prediction 1: 2 yr experience, 9 test score, 6 interview score
pred1 = reg.predict([[2, 9, 6]])
print(f"Candidate 1 Salary with 2 yr experience, 9 test score and 6 interview score is: ${pred1[0]:.2f}")

# Prediction 2: 12 yr experience, 10 test score, 10 interview score
pred2 = reg.predict([[12, 10, 10]])
print(f"Candidate 2 Salary with 12 years experience, 10 test score and 10 interview score is: ${pred2[0]:.2f}")
print("----------------------------------------------------------------------------------------------------------------------------------------------------------------")
```
